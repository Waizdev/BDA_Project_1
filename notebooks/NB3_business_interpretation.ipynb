```python
# ═══════════════════════════════════════════════════════════════
# NB03 — BUSINESS INTERPRETATION (SINGLE CELL)
# ═══════════════════════════════════════════════════════════════
from google.colab import drive, files
drive.mount('/content/drive/')

import pandas as pd, numpy as np

df = pd.read_csv("/content/drive/MyDrive/social_media_cleaned.csv")
print(f"Loaded: {df.shape}")

metrics = ["likes","retweets_shares","comments"]
plat = df.groupby("platform").agg(n=("post_id","count"),
    median_likes=("likes","median"),
    median_shares=("retweets_shares","median"),
    median_comments=("comments","median")).round(2).sort_values("median_likes", ascending=False)
sent = df.groupby("sentiment")[metrics].median().reindex(["Positive","Neutral","Negative"])
day_order = ["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
day = df.groupby("day_of_week").agg(n=("post_id","count"),
    median_likes=("likes","median")).reindex(day_order).round(2)

t1, t2 = plat.index[0], plat.index[1]
p_l, n_l = sent.loc["Positive","likes"], sent.loc["Negative","likes"]
bd = day["median_likes"].idxmax()

# ── FINDINGS BLOCKS ──
report = f"""# NB03 — BUSINESS INTERPRETATION

## Finding 1: Platform Effect on Engagement
**Finding:** {t1} posts achieve the highest median engagement.
**Evidence:** {t1} → median {plat.loc[t1,'median_likes']} likes, {plat.loc[t1,'median_shares']} shares, {plat.loc[t1,'median_comments']} comments (n={plat.loc[t1,'n']}). {t2} → median {plat.loc[t2,'median_likes']} likes (n={plat.loc[t2,'n']}). Delta = {plat.loc[t1,'median_likes']-plat.loc[t2,'median_likes']:+.1f} likes.
**Action:** Test more {t1}-native content in next 4-week cycle. Compare against current {t2} baseline using median likes.
**Limitation:** No reach/impressions → engagement rate cannot be computed. Volume effects uncontrolled.

## Finding 2: Sentiment Effect
**Finding:** Positive-sentiment posts show higher median likes than negative.
**Evidence:** Positive median = {p_l:.1f}, Neutral = {sent.loc['Neutral','likes']:.1f}, Negative = {n_l:.1f}. Delta (Pos − Neg) = {p_l - n_l:+.1f} likes.
**Action:** Shift content tone toward positive/value-led framing. Track sentiment of posted content.
**Limitation:** Correlation ≠ causation. Sentiment labels were pre-supplied by the dataset — provenance unknown.

## Finding 3: Day-of-Week Effect
**Finding:** {bd} shows the highest median likes.
**Evidence:** Best day = {bd} ({day.loc[bd,'median_likes']} median likes, n={int(day.loc[bd,'n'])}). Worst = {day['median_likes'].idxmin()} ({day['median_likes'].min()}, n={int(day.loc[day['median_likes'].idxmin(),'n'])}).
**Action:** Schedule high-priority posts on {bd}. Validate with 4-week A/B test.
**Limitation:** Day-of-week coverage uneven — small n on some days biases the result.

## Marketing Recommendation
Based on {len(df)} cleaned posts:
1. Prioritize **{t1}** for content investment (highest median likes = {plat.loc[t1,'median_likes']}).
2. Lean into **positive-sentiment** posts (Δ={p_l-n_l:+.1f} vs negative).
3. Schedule high-stakes posts on **{bd}**.

Measurement plan: 4-week controlled test — one post per platform per week, same content. Compare median likes + shares. Track by platform × day × sentiment to validate on the brand's own audience.

## Limitations (top-level)
- No reach or impressions → engagement RATE not computable
- followers_count missing for {df['followers_count'].isna().sum()} posts
- Dataset is NOT Pakistani brand data → findings are general, not Pakistan-specific
- Sentiment labels pre-supplied (provenance unknown)
- Final N = {len(df)} rows after cleaning

## Analytics Preview (Step 9 — Conceptual)
- **Classification:** Predict high/low engagement (label = median split). ⚠️ Data leakage guardrail: do not use `likes` in both label and features. Train/test split mandatory.
- **Clustering:** Group posts by standardized (likes, shares, comments) — reveals content archetypes.
- **Regression:** Predict likes from platform + sentiment + followers_count. Use log(likes) for right-skew. Train/test split.
- **Exit question:** Which technique best answers YOUR marketing question? (Regression fits best here — quantifies which factors predict engagement.)

## Next Step
Add paid-campaign data (spend, impressions, clicks, conversions) to enable ROAS analysis.
"""

print(report)

with open("/content/nb03_report.md", "w") as f:
    f.write(report)

files.download("/content/nb03_report.md")
print("\n✅ NB03 COMPLETE — nb03_report.md downloaded")
```
